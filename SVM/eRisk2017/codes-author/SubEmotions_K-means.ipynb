# Load words from EmoLex

```python
import os
import re

def get_texts_from_dir(cat_dir):
    texts = []
    f_names = []
    DIR = cat_dir

    for f_name in sorted(os.listdir(DIR)):
        f_path = os.path.join(DIR, f_name)
        f = open(f_path, "r")
        texts += [f.read()]
        f_names += [f_name]
        f.close()
    print("%d files loaded from %s" % (len(texts), cat_dir))
    return texts, f_names
```

```python
emo_words, emotions = get_texts_from_dir("./emotions_ext")
print(emotions)
print(len(emotions),len(emo_words))

em_words = []

for i in emo_words:
    em_words.append(i.split("\n"))
print(type(emo_words[0]),len(emo_words[0]))
print(type(em_words[0]),len(em_words[0]))
```

# Create clusters from EmoLex

```python
def convertToVec(words):
    X = []
    for i in words:
        X.append(wv[i])
    return X
```

```python
from sklearn.cluster import KMeans
import numpy as np

def cluster_KM(X,emotion,n):
    
    X = np.array(X)
    print("Creating clusters.. ")
    clustering = KMeans(n_clusters=n).fit(X)
    
    centers = clustering.cluster_centers_
    labels = clustering.labels_
    
    #print(labels)
    print("Numbers of clusters created of",emotion[7:-8],len(centers))
    return centers, labels
```

```python
X1 = convertToVec(em_words[0])
X2 = convertToVec(em_words[1])
X3 = convertToVec(em_words[2])
X4 = convertToVec(em_words[3])
X5 = convertToVec(em_words[4])
X6 = convertToVec(em_words[5])
X7 = convertToVec(em_words[6])
X8 = convertToVec(em_words[7])
X9 = convertToVec(em_words[8])
X10 = convertToVec(em_words[9])

print("Lenght of vocabulary: ",len(X1),len(X2),len(X3),len(X4),len(X5),
      len(X6),len(X7),len(X8),len(X9),len(X10))
#print(X1[0])

#num_clusters
n = 300 # emotions
m = 800 # sentiments

#clusters, labels // word vectors,name of the emotion, num_clusters
centers1, labels1 = cluster_KM(X1,emotions[0],444)
centers2, labels2 = cluster_KM(X2,emotions[1],395)
centers3, labels3 = cluster_KM(X3,emotions[2],367)
centers4, labels4 = cluster_KM(X4,emotions[3],488)
centers5, labels5 = cluster_KM(X5,emotions[4],318)
centers6, labels6 = cluster_KM(X6,emotions[5],818) # sentiment
centers7, labels7 = cluster_KM(X7,emotions[6],740) # sentiment
centers8, labels8 = cluster_KM(X8,emotions[7],395)
centers9, labels9 = cluster_KM(X9,emotions[8],274)
centers10, labels10 = cluster_KM(X10,emotions[9],383)

clusters = np.concatenate((centers1, centers2, centers3, centers4, centers5, centers6,
                          centers7, centers8, centers9, centers10), axis=0)
```
